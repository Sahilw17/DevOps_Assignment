# Proxie Studio DevOps Evaluation - Robot Explorer Python Bridge

This notebook demonstrates live bidirectional communication between a hosted static Three.js simulation and a local Python environment.

### Key Capabilities Demonstrated:
1. **Live State Observation**: Streaming continuous 60Hz telemetry (X, Z, heading, FPS) from a hosted web page without scraping screenshots.
2. **Programmatic Control**: Sending motion commands and executing closed-loop autonomous navigation from Python.
3. **Telemetry Analysis**: Capturing state trajectories for data science / robotics evaluation.

```python
# 1. Initialize RobotClient and connect to WebSocket Bridge
import asyncio
import time
from python_bridge.robot_client import RobotClient, RobotCommand

client = RobotClient("ws://127.0.0.1:8765")
await client.connect()
print(f"Connected to Bridge: {client.is_connected}")
```

```python
# 2. Inspect Live Robot Telemetry Stream
await asyncio.sleep(1.0) # Allow telemetry frames to stream

print(f"Robot Position: X = {client.state.x:.2f}, Z = {client.state.z:.2f}")
print(f"Robot Heading:  {client.state.rotation_deg:.1f} deg ({client.state.rotation_y:.3f} rad)")
print(f"Stream Rate:    {client.state.fps:.1f} FPS")
print(f"Origin Host:    {client.state.page_url}")
```

```python
# 3. Programmatic Motion Control
# Command the robot to drive forward and rotate
print("Driving forward for 2.0 seconds...")
await client.drive_for(duration=2.0, forward=True, run=True)

print("Rotating left for 1.0 second...")
await client.drive_for(duration=1.0, left=True)

print(f"New Position: X = {client.state.x:.2f}, Z = {client.state.z:.2f}")
```

```python
# 4. Autonomous Waypoint Navigation
# Navigate directly to coordinate (30.0, 30.0)
print("Navigating to waypoint (30.0, 30.0)...")
success = await client.goto(target_x=30.0, target_z=30.0, tolerance=2.5, max_seconds=20.0)
print(f"Navigation Success: {success}")
print(f"Final Position: X = {client.state.x:.2f}, Z = {client.state.z:.2f}")
```

```python
# 5. Clean Disconnect
await client.disconnect()
print("Disconnected from bridge.")
```
